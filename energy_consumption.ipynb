```python
from sklearn.model_selection import train_test_split

# Separate features (X) and target (y)
X = df.drop(columns=["EnergyConsumption", "Timestamp"])
y = df["EnergyConsumption"]

# Split the dataset: 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=40
)

# Check the shapes
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder

# Select categorical columns
categorical_features = [  "HVACUsage",  "LightingUsage",    "DayOfWeek", "Holiday"]

# Create encoder
encoder = OneHotEncoder(handle_unknown="ignore")

# Fit on training data and encode
X_train_encoded = encoder.fit_transform(X_train[categorical_features])

# Encode test data using the same encoder
X_test_encoded = encoder.transform(
    X_test[categorical_features]
)

print(X_train_encoded.shape)
print(X_test_encoded.shape)
```

```python
from sklearn.preprocessing import StandardScaler

numerical_features = [
    col for col in X_train.columns
    if col not in categorical_features
]

scaler = StandardScaler()

X_train_num_scaled = scaler.fit_transform(
    X_train[numerical_features]
)

X_test_num_scaled = scaler.transform(
    X_test[numerical_features]
)
print("scaling done")
```

```python
df.sample(5)
```

```python
# See the encoded training data
print(encoder.get_feature_names_out(categorical_features))
print(X_train_encoded.toarray()[:5])
```

```python

from scipy.sparse import hstack, csr_matrix

# Select the numerical columns

# Combine encoded categorical columns with numerical columns
X_train_final = hstack([
    X_train_encoded,
    csr_matrix(X_train_num_scaled)]).tocsr()

X_test_final = hstack([
    X_test_encoded,
    csr_matrix(X_test_num_scaled)]).tocsr()

print("Training data shape:", X_train_final.shape)
print("Test data shape:", X_test_final.shape)
```

```python

from sklearn.linear_model import LinearRegression

# Create the model
model = LinearRegression()

# Train the model
model.fit(X_train_final, y_train)

print("Model trained successfully! wow😀 ")
```

```python

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

# Predict energy consumption
y_pred = model.predict(X_test_final)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Baseline Model Results")
print("----------------------")
print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```
